# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
import logging

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[logging.StreamHandler()]
)
logger = logging.getLogger('AgentLogger')

# 🛠️ TOOL 3: Search (Bonus)
def search_tool(query: str) -> str:
    """Simulate a search tool."""
    logger.info(f"Using Search Tool for: {query}")
    return f"Search results for '{query}': Found related information on the web."
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
def agent(query: str):
    query_lower = query.lower()

    if "calculate" in query_lower:
        # Basic extraction: remove the word 'calculate' to get the expression
        expression = query_lower.replace("calculate", "").strip()
        result = calculator(expression)
        return {
            "type": "calculation",
            "result": result
        }

    elif "keywords" in query_lower:
        # Extract keywords using the tool
        result = extract_keywords(query)
        return {
            "type": "keywords",
            "result": result
        }

    else:
        # General direct response
        return {
            "type": "general",
            "result": f"I understand you are asking about: {query}"
        }
````

````python
def agent(query: str):
    logger.info(f"Received query: {query}")
    query_lower = query.lower()

    try:
        # Improved Routing with more patterns
        if any(x in query_lower for x in ["calculate", "math", "plus", "minus", "*", "/"]):
            # Cleaning expression
            expression = query_lower.replace("calculate", "").strip()
            result = calculator(expression)
            return {"type": "calculation", "result": result}

        elif "keywords" in query_lower or "extract" in query_lower:
            result = extract_keywords(query)
            return {"type": "keywords", "result": result}

        elif "search" in query_lower or "who" in query_lower or "tell me about" in query_lower:
            result = search_tool(query)
            return {"type": "search", "result": result}

        else:
            return {
                "type": "general",
                "result": f"I understand you are asking about: {query}"
            }
    except Exception as e:
        logger.error(f"Error processing query: {e}")
        return {"type": "error", "result": "Something went wrong inside the agent logic."}
````
